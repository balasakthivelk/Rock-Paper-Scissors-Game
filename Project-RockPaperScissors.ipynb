# Rock / Paper / Scissors Game - V3 (Counts the total wins)

```python
#!pip install numpy
def RockPaperScissors():
    Options=["Rock", "Paper", "Scissors"]
    dict1={"Person":0, "Computer":0, "Draw":0}
    select_options(Options, dict1)

def select_options(Options, dict1):
    import random
    Person=input("Type your option (Rock / Paper / Scissors) or (Stop) - to end game : ")
    Person = Person.title()
    if Person == "Stop":
        print("You Stopped the game")
        return        
    if Person not in Options:
        print(f"Your input '{Person}', doesnt match (Rock / Paper / Scissors) \n ")
        select_options(Options, dict1)
    else:    
        Computer = random.choice(Options)
                
        print("Your choice: ", Person)
        print("Computer choice: ", Computer)

        check_winner(Person , Computer, dict1)
        select_options(Options, dict1)

def check_winner(Person , Computer, dict1):
    if Computer == Person:
        print("It's a draw \n ")
        dict1["Draw"]=dict1["Draw"]+1
    elif Computer == "Rock" and  Person == "Paper":
        print("Person Wins \n ")
        dict1["Person"]=dict1["Person"]+1
    elif Computer == "Rock" and Person == "Scissors":
        print("Computer Wins \n ")  
        dict1["Computer"]=dict1["Computer"]+1
    elif Computer == "Paper" and Person == "Scissors":
        print("Person Wins \n ")  
        dict1["Person"]=dict1["Person"]+1
    elif Computer == "Paper" and Person == "Rock":
        print("Computer Wins \n ") 
        dict1["Computer"]=dict1["Computer"]+1
    elif Computer == "Scissors" and Person == "Rock":
        print("Person Wins \n ") 
        dict1["Person"]=dict1["Person"]+1
    elif Computer == "Scissors" and Person == "Paper":
        print("Computer Wins \n ") 
        dict1["Computer"]=dict1["Computer"]+1
    print("Total: "+ 
          "You Win " + str(dict1["Person"]) + " time(s), "
          "Computer Win " + str(dict1["Computer"]) + " time(s), "          
          "Draw by " + str(dict1["Draw"]) + " time(s). \n ")    
    #print(f"Total: You Win '{dict1["Person"]}' times, Computer Win '{dict1["Computer"]}' times, Draw by '{dict1["Draw"]}' times.")   
RockPaperScissors()
```
